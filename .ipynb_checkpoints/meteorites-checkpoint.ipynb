- **Data I/O:** You start by loading the data using pd.read_csv(). Later, you will export their filtered findings using .to_excel() or .to_json() for a web team to use.


```python
df = pd.read_csv("meteorites.csv")

df.to_excel("heavy_meteorites.xlsx", index=False)
print("Saved Excel file")
df.to_json("heavy_meteorites.json", orient="records")
print("Saved JSON file")
```

- **Initial Data Inspection:** Before doing anything, you must run .info() and .describe() to understand the sheer volume of data, use .head() to see the layout, and use .value_counts() to see which regions have the most frequent seismic activity.


```python
df.info()
df.describe()
print(df.head())

# value_counts
region = df["region"]
print(region.value_counts())
```

**Indexing and Selection:**
- Column/Label/Integer: Use dictionary-style selection to grab the fall_status column (whether it was seen falling or just found later). Use .iloc[] to extract the top 5 heaviest meteorites based on their sorted row position.


```python
fall_status = df["fall_status"]
print(fall_status.value_counts())

df_sorted = df.sort_values(by="mass_grams", ascending=False)
top_5_heaviest = df_sorted.iloc[:5]
print(f"Top 5 heaviest Use .iloc[]:\n{top_5_heaviest}")
```

- Filtering: Apply multiple boolean conditions using & and | to find meteorites that are both greater than 50,000 grams in mass and were discovered in a specific region (e.g., "Antarctica" or "Sahara").

```python
print("mass_grams > 50000:\n")
filtered = df[
    (df["mass_grams"] > 50000) &
    ((df["region"] == "Antarctica") | (df["region"] == "Sahara"))
]
print(filtered)
```
